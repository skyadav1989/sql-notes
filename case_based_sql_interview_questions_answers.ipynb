## 1. Create Tables

````python
CREATE TABLE customers (
    customer_id INT PRIMARY KEY,
    customer_name VARCHAR(100),
    city VARCHAR(50),
    signup_date DATE
);

CREATE TABLE products (
    product_id INT PRIMARY KEY,
    product_name VARCHAR(100),
    category VARCHAR(50),
    price DECIMAL(10,2)
);

CREATE TABLE orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    order_date DATE,
    status VARCHAR(20),
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
);

CREATE TABLE order_items (
    order_item_id INT PRIMARY KEY,
    order_id INT,
    product_id INT,
    quantity INT,
    unit_price DECIMAL(10,2),
    FOREIGN KEY (order_id) REFERENCES orders(order_id),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);
````

## SQL Interview Practice

Each question contains:
1. Business requirement
2. SQL query
3. Interview concept

For revenue questions, the business rule is:

`revenue = quantity × unit_price`

Unless explicitly stated otherwise, revenue questions use **Completed** orders.


### Q1. Find all customers from Delhi.

**Concept:** WHERE filtering

````python
SELECT *
FROM customers
WHERE city = 'Delhi';
````

### Q2. Find products costing more than ₹10,000.

**Concept:** WHERE + comparison

````python
SELECT *
FROM products
WHERE price > 10000;
````

### Q3. Find all completed orders.

**Concept:** Filtering

````python
SELECT *
FROM orders
WHERE status = 'Completed';
````

### Q4. Find customers who signed up after March 2024.

**Concept:** Date filtering

````python
SELECT *
FROM customers
WHERE signup_date > '2024-03-31';
````

### Q5. Display Electronics products.

**Concept:** Filtering

````python
SELECT *
FROM products
WHERE category = 'Electronics';
````

### Q6. Find the total number of orders.

**Concept:** COUNT

````python
SELECT COUNT(*) AS total_orders
FROM orders;
````

### Q7. Find the number of completed orders.

**Concept:** COUNT + WHERE

````python
SELECT COUNT(*) AS completed_orders
FROM orders
WHERE status = 'Completed';
````

### Q8. Calculate the value of every order.

**Concept:** GROUP BY + SUM

````python
SELECT
    order_id,
    SUM(quantity * unit_price) AS order_value
FROM order_items
GROUP BY order_id;
````

### Q9. Calculate revenue from completed orders only.

**Concept:** JOIN + aggregation

````python
SELECT
    SUM(oi.quantity * oi.unit_price) AS total_revenue
FROM order_items oi
JOIN orders o
    ON oi.order_id = o.order_id
WHERE o.status = 'Completed';
````

### Q10. Find total revenue generated by each customer.

**Concept:** Multiple JOINs + GROUP BY

````python
SELECT
    c.customer_id,
    c.customer_name,
    SUM(oi.quantity * oi.unit_price) AS total_revenue
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
JOIN order_items oi
    ON o.order_id = oi.order_id
WHERE o.status = 'Completed'
GROUP BY c.customer_id, c.customer_name;
````

### Q11. Find customers who placed at least 2 orders.

**Concept:** HAVING

````python
SELECT
    c.customer_id,
    c.customer_name,
    COUNT(o.order_id) AS order_count
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
HAVING COUNT(o.order_id) >= 2;
````

### Q12. Find the most expensive product.

**Concept:** ORDER BY + LIMIT

````python
SELECT *
FROM products
ORDER BY price DESC
LIMIT 1;
````

### Q13. Find products that have never been ordered.

**Concept:** LEFT JOIN + anti-join

````python
SELECT
    p.product_id,
    p.product_name
FROM products p
LEFT JOIN order_items oi
    ON p.product_id = oi.product_id
WHERE oi.product_id IS NULL;
````

### Q14. Find customers who have never placed an order.

**Concept:** LEFT JOIN + IS NULL

````python
SELECT
    c.customer_id,
    c.customer_name
FROM customers c
LEFT JOIN orders o
    ON c.customer_id = o.customer_id
WHERE o.customer_id IS NULL;
````

### Q15. Find the highest-spending customer.

**Concept:** Aggregation + ranking

````python
SELECT
    c.customer_id,
    c.customer_name,
    SUM(oi.quantity * oi.unit_price) AS total_spent
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
JOIN order_items oi ON o.order_id = oi.order_id
WHERE o.status = 'Completed'
GROUP BY c.customer_id, c.customer_name
ORDER BY total_spent DESC
LIMIT 1;
````

### Q16. Find the second-highest spending customer.

**Concept:** CTE + DENSE_RANK

````python
WITH customer_revenue AS (
    SELECT
        c.customer_id,
        c.customer_name,
        SUM(oi.quantity * oi.unit_price) AS total_spent
    FROM customers c
    JOIN orders o ON c.customer_id = o.customer_id
    JOIN order_items oi ON o.order_id = oi.order_id
    WHERE o.status = 'Completed'
    GROUP BY c.customer_id, c.customer_name
),
ranked_customers AS (
    SELECT *,
           DENSE_RANK() OVER (ORDER BY total_spent DESC) AS rnk
    FROM customer_revenue
)
SELECT *
FROM ranked_customers
WHERE rnk = 2;
````

### Q17. Find the best-selling product by quantity.

**Concept:** JOIN + SUM + ranking

````python
SELECT
    p.product_id,
    p.product_name,
    SUM(oi.quantity) AS total_quantity
FROM products p
JOIN order_items oi ON p.product_id = oi.product_id
JOIN orders o ON oi.order_id = o.order_id
WHERE o.status = 'Completed'
GROUP BY p.product_id, p.product_name
ORDER BY total_quantity DESC
LIMIT 1;
````

### Q18. Find revenue generated by each product.

**Concept:** Product revenue

````python
SELECT
    p.product_id,
    p.product_name,
    SUM(oi.quantity * oi.unit_price) AS revenue
FROM products p
JOIN order_items oi ON p.product_id = oi.product_id
JOIN orders o ON oi.order_id = o.order_id
WHERE o.status = 'Completed'
GROUP BY p.product_id, p.product_name
ORDER BY revenue DESC;
````

### Q19. Find revenue by product category.

**Concept:** GROUP BY category

````python
SELECT
    p.category,
    SUM(oi.quantity * oi.unit_price) AS revenue
FROM products p
JOIN order_items oi ON p.product_id = oi.product_id
JOIN orders o ON oi.order_id = o.order_id
WHERE o.status = 'Completed'
GROUP BY p.category
ORDER BY revenue DESC;
````

### Q20. Find the top 3 customers by revenue.

**Concept:** Top-N analysis

````python
SELECT
    c.customer_id,
    c.customer_name,
    SUM(oi.quantity * oi.unit_price) AS revenue
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
JOIN order_items oi ON o.order_id = oi.order_id
WHERE o.status = 'Completed'
GROUP BY c.customer_id, c.customer_name
ORDER BY revenue DESC
LIMIT 3;
````

### Q21. Find customers spending more than the average customer spending.

**Concept:** CTE + subquery

````python
WITH customer_spending AS (
    SELECT
        c.customer_id,
        c.customer_name,
        SUM(oi.quantity * oi.unit_price) AS spending
    FROM customers c
    JOIN orders o ON c.customer_id = o.customer_id
    JOIN order_items oi ON o.order_id = oi.order_id
    WHERE o.status = 'Completed'
    GROUP BY c.customer_id, c.customer_name
)
SELECT *
FROM customer_spending
WHERE spending > (SELECT AVG(spending) FROM customer_spending);
````

### Q22. Find the latest order of every customer.

**Concept:** ROW_NUMBER + PARTITION BY

````python
WITH ranked_orders AS (
    SELECT
        o.*,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id
            ORDER BY order_date DESC
        ) AS rn
    FROM orders o
)
SELECT *
FROM ranked_orders
WHERE rn = 1;
````

### Q23. Find the first order of every customer.

**Concept:** ROW_NUMBER

````python
WITH ranked_orders AS (
    SELECT
        o.*,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id
            ORDER BY order_date
        ) AS rn
    FROM orders o
)
SELECT *
FROM ranked_orders
WHERE rn = 1;
````

### Q24. Find customers who placed orders in both July and August.

**Concept:** Conditional aggregation

````python
SELECT
    c.customer_id,
    c.customer_name
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
WHERE o.status = 'Completed'
GROUP BY c.customer_id, c.customer_name
HAVING SUM(MONTH(o.order_date) = 7) > 0
   AND SUM(MONTH(o.order_date) = 8) > 0;
````

### Q25. Find monthly revenue.

**Concept:** Date grouping

````python
SELECT
    DATE_FORMAT(o.order_date, '%Y-%m') AS month,
    SUM(oi.quantity * oi.unit_price) AS revenue
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
WHERE o.status = 'Completed'
GROUP BY DATE_FORMAT(o.order_date, '%Y-%m')
ORDER BY month;
````

### Q26. Find the highest revenue month.

**Concept:** CTE + Top-1

````python
WITH monthly_revenue AS (
    SELECT
        DATE_FORMAT(o.order_date, '%Y-%m') AS month,
        SUM(oi.quantity * oi.unit_price) AS revenue
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
    WHERE o.status = 'Completed'
    GROUP BY DATE_FORMAT(o.order_date, '%Y-%m')
)
SELECT *
FROM monthly_revenue
ORDER BY revenue DESC
LIMIT 1;
````

### Q27. Rank customers based on spending.

**Concept:** RANK

````python
WITH customer_spending AS (
    SELECT
        c.customer_id,
        c.customer_name,
        SUM(oi.quantity * oi.unit_price) AS spending
    FROM customers c
    JOIN orders o ON c.customer_id = o.customer_id
    JOIN order_items oi ON o.order_id = oi.order_id
    WHERE o.status = 'Completed'
    GROUP BY c.customer_id, c.customer_name
)
SELECT *,
       RANK() OVER (ORDER BY spending DESC) AS customer_rank
FROM customer_spending;
````

### Q28. Find the top-selling product in each category.

**Concept:** PARTITION BY + RANK

````python
WITH product_sales AS (
    SELECT
        p.category,
        p.product_id,
        p.product_name,
        SUM(oi.quantity) AS total_quantity
    FROM products p
    JOIN order_items oi ON p.product_id = oi.product_id
    JOIN orders o ON oi.order_id = o.order_id
    WHERE o.status = 'Completed'
    GROUP BY p.category, p.product_id, p.product_name
),
ranked_products AS (
    SELECT *,
           RANK() OVER (
               PARTITION BY category
               ORDER BY total_quantity DESC
           ) AS rnk
    FROM product_sales
)
SELECT *
FROM ranked_products
WHERE rnk = 1;
````

### Q29. Calculate each customer's percentage contribution to total revenue.

**Concept:** Window aggregate

````python
WITH customer_revenue AS (
    SELECT
        c.customer_id,
        c.customer_name,
        SUM(oi.quantity * oi.unit_price) AS revenue
    FROM customers c
    JOIN orders o ON c.customer_id = o.customer_id
    JOIN order_items oi ON o.order_id = oi.order_id
    WHERE o.status = 'Completed'
    GROUP BY c.customer_id, c.customer_name
)
SELECT
    customer_id,
    customer_name,
    revenue,
    ROUND(
        revenue * 100.0 / SUM(revenue) OVER (),
        2
    ) AS revenue_percentage
FROM customer_revenue
ORDER BY revenue DESC;
````

### Q30. Find customers who have completed orders but never had a cancelled order.

**Concept:** Conditional aggregation + HAVING

````python
SELECT
    c.customer_id,
    c.customer_name
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
HAVING COUNT(CASE WHEN o.status = 'Completed' THEN 1 END) > 0
   AND COUNT(CASE WHEN o.status = 'Cancelled' THEN 1 END) = 0;
````

# Key Interview Patterns

## 1. Find records with no match

```sql
LEFT JOIN ...
WHERE right_table.id IS NULL
```

## 2. Filter aggregated results

```sql
GROUP BY ...
HAVING COUNT(*) > 1
```

## 3. Top N

```sql
ORDER BY metric DESC
LIMIT N
```

## 4. Latest row per group

```sql
ROW_NUMBER() OVER (
    PARTITION BY group_column
    ORDER BY date_column DESC
)
```

## 5. Ranking

```sql
RANK() OVER (ORDER BY metric DESC)
```

## 6. Compare with previous row

```sql
LAG(metric) OVER (
    ORDER BY date_column
)
```

## 7. Running total

```sql
SUM(metric) OVER (
    ORDER BY date_column
)
```

## 8. Top record in every group

```sql
RANK() OVER (
    PARTITION BY group_column
    ORDER BY metric DESC
)
```
